```python
# "with" keyword is used for unmanaged data
with open('myfile.txt') as my_file:
    file_read = my_file.read()
```

```python
file_read
```

```python
# mode specifies permisions (mode ="r") read mode
with open("myfile.txt", mode = "r") as my_newfile1:
    readfile = my_newfile1.read()
```

```python
readfile
```

```python
# mode specifies permisions (mode = "w") write mode / overwrite
with open("myfile.txt", mode = "w") as my_newfile2:
    my_newfile2.write("test")
```

```python
# not allowed to read
my_newfile2
```

```python
# mode specifies permisions (mode = "a") append mode / add
with open("myfile.txt", mode = "a") as my_newfile3:
    my_newfile3.write("test 3")
```

```python
# reading back appended text
with open("myfile.txt", mode = "r") as my_newfile3:
    print(my_newfile3.read())
```

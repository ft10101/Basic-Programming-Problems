# Dunder Methods / Special methods

```python
class fruits():
    
    def __init__(self,name,calories):
        self.name = name
        self.calories = calories
        
    def __str__(self): 
    # method for printing instances
        return f"{self.name} has {self.calories}"
    
    def __len__(self):
        return self.calories
```

```python
my_fruits = fruits('apple',10)
```

```python
my_fruits.calories
```

```python
print(my_fruits)
```

```python
str(my_fruits)
```

# dividing list by functions

```python
def loop(mylist):
    print("inside function")
    for x in mylist:
        print(mylist[x-1]/2)
    return
```

```python
loop(my_list)
```

```python
def divide(mylist):
    new_list = []
    for c in mylist:
        new_list.append(c/2)
    print(new_list)
```

```python
divide(my_list)
```

```python
for xi in my_list:
    print(divi(xi))
```

```python
def string_control(string):
    return "metallica" in string
```

```python
string_control("metallica and sona")
```

```python
my_artist = ["metallica","medona","sona","metallica2"]
```

```python
list(map(string_control,my_artist))
```

#### get name of artist that has name metallica in their name, 
#### elements that are true in

*list(map(string_control,my_artist))*

```python
q = []
q = list(map(string_control,my_artist))
print("first element in q is :", q[0],",, first artist is:", my_artist[0])
print("second element in q is :", q[1],",,second artist is: ", my_artist[1])
print("third element in q is :", q[2],",,third artist is: ", my_artist[2])
print("fourth element in q is :", q[3],",,fourth artist is: ", my_artist[3])
```

```python
for a  in range(len(q)):
    print("artist is: ", my_artist[a])
    print("a is :",a)
```

```python
for a in range(len(q)):
    print(q[a])
    print(a)
```

```python
new_artist = []
for a in range(len(q)):
    if q[a] is True:
        new_artist.append(my_artist[a])
    else:
        print("not metallica")
```

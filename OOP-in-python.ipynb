```python
class Musician(): # declaration of class

    job = "musician" # default valued attribute
    
    # initializer funtion
    def __init__(self, name, age,instrument):
        self.name_attribute = name
        self.age_attribute = age
        self.instrument_attribute = instrument
        
    #defining a method
    def sing(self):
        print(f"singing with: {self.instrument_attribute}")
```

```python
my_musician = Musician("james", 40,"guitar")
```

```python
my_musician.name_attribute
```

```python
my_musician.instrument_attribute
```

```python
my_musician.sing()
```

# practical example

```python
class DogYears():
    
    year_factor = 7
    
    def __init__(self,age=3):
        self.age = age
        self.multiplied_age = age * 7
        
    def calculate(self):
        return self.age * DogYears.year_factor
```

```python
my_dog = DogYears()
```

```python
my_dog.calculate()
```

```python
my_dog.multiplied_age
```

# Inheritance

```python
class class1():
    
    def __init__(self):
        print("class1 created")
        
    def method1(self):
        print("method 1")
        
    def method2(self):
        print("method 2")
```

```python
myInstance = class1()
```

```python
myInstance.method1()
```

```python
myInstance.method2()
```

```python
class class2(class1):
    # inheritance
    def __init__(self):
        class1.__init__(self)
        print("class 2 created")
        
    def method3(self):
        print("method 3")
        
    # override
    def method1(self):
        print("method 1 overiden")
```

```python
myInstance2 = class2()
```

```python
myInstance2.method1()
```

```python
myInstance2.method3()
```

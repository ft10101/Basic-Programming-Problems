## Scope of variables

```python
number = 10 # -- global variable

def multiply(num):
    number = 5  # --local variable
    return num*number
```

```python
multiply(10)
```

```python
number
```

```python
a = 10
a = 5 
print(a)
```

# LEGB
## L -> local
## E -> Enclosing
## G -> Global
## B -> Build-in

```python
my_string = "atil"
# -------------------- global level
def my_func():
    # ------------------ enclosed level
    my_string = "james"
    def my_func2():
    # ---------------------- local level
        my_string = "mano"
        print(my_string)
    my_func2()
```

```python
my_func()
```

```python
my_string
```

```python
y = 10
def fun(y):
    print("y before changing is:",y)
    y = 5
    print("y after changing is:",y)
```

```python
fun(y)
```

```python
y
```

```python
y = 10
def fun(y):
    print("y before changing is:",y)
    y = 5
    print("y after changing is:",y)
    return y

y = fun(y) # changing global variable
```

```python
y
```

```python
x = 10
def func():
    global x  #takes global value of x
    x = 5
    print(x)
```

```python
func()
```

```python
x
```

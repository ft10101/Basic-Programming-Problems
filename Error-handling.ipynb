# try, except, finally

```python
while True:
# error handling using try and except
    try:
        a = int(input("enter number: "))
    except:
        print("enter the number!")
        continue
    # if program doesn't go towards exception then else is executed
    else:
        print("ok")
        break
    # finally is executed all the time
    finally:
        print("finally is executed always")
```
